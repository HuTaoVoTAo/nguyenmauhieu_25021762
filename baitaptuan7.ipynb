```python
#W7A1
#viet ham tim kiem nhi phan
def binary_search(l,r,x,a):
    '''
        ham tim kiem nhi phan:
            *l,r la dau cuoi cua phan dang xet
            *x la con can tim 
            *a la mang dang xet
        tim vi tri cua x trong mang tang dan
    '''
    mid=int((l+r)/2)
    if (l==r) or (a[mid]==x):
        return mid
    elif (a[mid]<x):
        return binary_search(mid+1,r,x,a)
    else:
        return binary_search(l,mid-1,x,a)
n=[int(i) for i in input().split()]
x=int(input())
t=binary_search(0,len(n)-1,x,n)
if x==n[t]:
    print(t)
else:
    print(-1)

```

```python
#W7A2
def count(x,a):
    '''
        ham dem so trong mang:
            *x la so can dem
            *a la mang dang xet
        dem so luong so x xuat hien trong a
    '''
    d=0
    for i in a:
        if i==x:
            d+=1
    return d
n=[int(i) for i in input().split()]
x=int(input())
print(count(x,n))
```

```python
#W7A3
def selec_sort(a):
    t=len(a)
    for i in range (0,t):
        pos=-1
        mi=1e9
        for j in range (i+1,t):
            if mi>a[j]:
                mi=a[j]
                pos=j
        if pos>0:
            a[i],a[j]=a[j],a[i]
    return a
def bubble(a):
    t=len(a)
    for i in range (0,t):
        for j in range (i+1,t):
            if a[i]>a[j]:
                a[i],a[j]=a[j],a[i]
    return a
n=[int(i) for i in input().split()]
print(selec_sort(n))
print(bubble(n))
```

```python
#W7A13
a=list(map(int,input().split()))
a.sort()
mx=1
cur=1
for i in range(len(a)):
    for j in range(i+1,len(a)):
        if (a[j]-a[i]==1) or (a[i]==[j]):
            cur+=1
            if cur>mx:
                mx=cur
        else:
            cur=1
print(mx)
```
