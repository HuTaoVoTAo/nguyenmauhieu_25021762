```python
#W7A16
dic={}
a=list(map(int,input().split()))
for i in a:
    if i not in dic:
        dic[i]=0=
    dic[i]+=1
ans=0
for i in a:
    if dic[i]>i:
        ans+=dic[i]-i
    elif dic[i]<i:
        ans+=dic[i]
    dic[i]=0
if ans:
    print("NO",ans)
else:
    print("YES")
```

```python
#w7a19
def merge(a,b):
    l,r=0,0
    ans=[]
    while l<len(a) and r<len(b):
        if a[l]<b[r]:
            ans.append(a[l])
            l+=1
        else:
            ans.append(b[r])
            r+=1
    a=ans+a[l::]+b[r::]
    for i in a:
        print(i,end=" ")
a=list(map(int,input().split()))
b=list(map(int,input().split()))
merge(a,b)
```

```python
#w7a27
a=input()
dic={'}':'{',']':'[','>':'<',')':'('}
stack=[]
for i in a:
    if i=='{' or i=='[' or i=='<' or i=='(':
        stack.append(i)
    else:
        if len(stack)>0 and stack[-1]==dic[i]:
            stack.pop()
        else:
            break
    #print(stack)
if len(stack)==0:
    print("YES")
else:
    print("NO")
```

```python
#W7A28
N = int(input())

nguyen_to = [1]*(N+1)
if N >= 1: nguyen_to[0] = nguyen_to[1] = 0

for i in range(2, int(N**0.5)+1):
    if nguyen_to[i]:
        for j in range(i*i, N+1, i):
            nguyen_to[j] = 0

tot_nhat = []
tam = []

for i in range(1, N):
    if i >= 4 and not nguyen_to[i]:
        tam.append(i)
    else:
        if len(tam) > len(tot_nhat):
            tot_nhat = tam[:]
        tam = []
if len(tam) > len(tot_nhat):
    tot_nhat = tam[:]

if not tot_nhat:
    print("Not found")
else:
    print(*tot_nhat)
```

```python
#W7A29
a = list(map(int, input().split()))

tot = []
tam = [a[0]]

for i in range(1, len(a)):
    if a[i] > a[i-1]:
        tam.append(a[i])
    else:
        if len(tam) > len(tot):
            tot = tam[:]
        tam = [a[i]]

if len(tam) > len(tot):
    tot = tam[:]

print(*tot)
```

```python
#W7A30
a = list(map(int, input().split()))
n = len(a)

dp = [1]*n     
truoc = [-1]*n 

for i in range(n):
    for j in range(i):
        if a[j] < a[i] and dp[j] + 1 > dp[i]:
            dp[i] = dp[j] + 1
            truoc[i] = j

i = dp.index(max(dp))

kq = []
while i != -1:
    kq.append(a[i])
    i = truoc[i]
print(*reversed(kq))
```

```python
#W7A31
s = list(input().strip())
n = len(s)

i = 0
while i < n-1:
    mn = min(s[i+1:])
    if mn < s[i]:
        j = ''.join(s).rfind(mn)
        if i == 0 and mn == '0':
            i += 1
            continue
        s[i], s[j] = s[j], s[i]
        break
    i += 1

print("".join(s))
```
