```python
#W6A19
def minmax(a):
    mi=1e9
    ma=-1e9
    posmi=-1
    posmx=-1
    for i in range(0,len(a)):
        if mi>a[i]:
            mi=min(mi,a[i])
            posmi=i
        if ma<a[i]:
            ma=max(ma,a[i])
            posma=i
    return mi,posmi,ma,posma
n=[int(i) for i in input().split()]
print(minmax(n))
```

```python
#W6A24
def new(a):
    fina=[]
    fi0=[]
    for i in a:
        if i!=0:
            fina.append(i)
        else:
            fi0.append(0)
    return(fi0+fina)
    
n=[int(i) for i in input().split()]
print(new(n))
```

```python
#W6A27
def find(a,k):
    ans=[]
    for i in range (0,len(a)):
        if a[i]<k:
            ans.append(i)
    if len(ans):
        return ans
    return -1
n=[int(i) for i in input().split()]
x=int(input())
print(find(n,x))
```
