```python
#w11a4
from pathlib import Path
WORKDIR = Path(".")
pat1=WORKDIR/"a.txt"
pat2=WORKDIR/"b.txt"
def areAnagrams(path1,path2):
    with open(path1,'r') as f1,open(path2,'r') as f:
        a=list(f1.readline().replace(" ",""))
        for i in range (0,len(a)):
            if a[i]>='A' and a[i]<='Z':
                a[i]=chr(ord(a[i])+31+1)
        a.sort()
        b=list(f.readline().replace(" ",""))
        for i in range (0,len(b)):
            if b[i]>='A' and b[i]<='Z':
                b[i]=chr(ord(b[i])+31+1)
        b.sort()
        if a==b:
            return "TRUE"
        return "FALSE"
print(areAnagrams(pat1,pat2))
```

```python
#w11a10
from pathlib import Path
WORKDIR = Path(".")
path = "a.txt"
path = WORKDIR / path
def getMoney (path: str) -> list[int]:
    with open(path,"r") as f:
        data = f.read().strip()
        if data == "":
            return []
        return list(map(int, data.split()))
def rob(houses: list[int]) -> int:
    prev2 = 0  
    prev1 = 0   

    for money in houses:
        cur = max(prev1, prev2 + money)
        prev2 = prev1
        prev1 = cur
    return prev1
houses = getMoney(path)
print(rob(houses))
```

```python
#w11a13
from pathlib import Path 
WORKDIR = Path(".")
path = WORKDIR/path
def moveZeroes(fileName: str) -> list:
    f = open(fileName)
    f.readline()
    a = list(map(int, f.read().split()))
    f.close()

    kq = []
    dem0 = 0

    for x in a:
        if x == 0:
            dem0 += 1
        else:
            kq.append(x)

    return kq + [0]*dem0
print(moveZeroes(input()))
```

```python
#w11a16
from pathlib import Path 
WORKDIR = Path(".")
path = WORKDIR/path
import math

path = input()
f = open(path)

dong = [x.strip() for x in f if x.strip() != ""]
f.close()

i = 0
while i < len(dong):
    loai = dong[i].split(":")[1].strip()
    ts = dong[i+1]
    i += 2

    if loai == "SQUARE":
        a = float(ts.split(":")[1])
        cv = 4 * a

    elif loai == "RECTANGLE":
        p = ts.replace("WIDTH:", "").replace("HEIGHT:", "").split()
        w, h = map(float, p)
        cv = 2 * (w + h)

    else:  # CIRCLE
        r = float(ts.split(":")[1])
        cv = 2 * math.pi * r

    print(f"{cv:.2f}")
```

```python
#w11a17
from pathlib import Path 
WORKDIR = Path(".")
path = WORKDIR/path
def productExceptSelf(path: str) -> list[int]:
    f = open(path)
    f.readline()      
    a = list(map(int, f.read().split()))
    f.close()

    n = len(a)
    res = [1]*n

    trai = 1
    for i in range(n):
        res[i] = trai
        trai *= a[i]

    phai = 1
    for i in range(n-1, -1, -1):
        res[i] *= phai
        phai *= a[i]

    return res
duong_dan = input()
kq = productExceptSelf(duong_dan)
print(*kq)
```

```python
#w11a18
from pathlib import Path 
WORKDIR = Path(".")
path = WORKDIR/path
import math

def estimatedTime(path: str):
    f = open(path)
    a = f.readline().split()
    b = f.readline().split()
    v = float(f.readline().split()[1])
    f.close()

    x1, y1 = float(a[2]), float(a[4])
    x2, y2 = float(b[2]), float(b[4])

    d = math.sqrt((x2-x1)**2 + (y2-y1)**2)
    return round(d / v, 2)
duong_dan = input()
print(estimatedTime(duong_dan))
```
