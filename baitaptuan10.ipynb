```python
#w9a1
def merge(a,b):
    l,r=0,0
    ans=[]
    while l<len(a) and r<len(b):
        if a[l]<b[r]:
            ans.append(a[l])
            l+=1
        else:
            ans.append(b[r])
            r+=1
    a=ans+a[l::]+b[r::]
    for i in a:
        print(i,end=" ")
a=list(map(int,input().split()))
b=list(map(int,input().split()))
merge(a,b)
```

```python
#W9A4
Q = int(input())

for _ in range(Q):
    a = input().strip()
    b = input().strip()

    n, m = len(a), len(b)
    dp = [[False]*(m+1) for _ in range(n+1)]
    dp[0][0] = True

    for i in range(n):
        for j in range(m+1):
            if not dp[i][j]: 
                continue

            if a[i].islower():
                dp[i+1][j] = True

            if j < m and a[i].upper() == b[j]:
                dp[i+1][j+1] = True

    print("YES" if dp[n][m] else "NO")
```

```python
#W9A7
def canPlaceFlowers(flowerbed, k):
    n = len(flowerbed)

    for i in range(n):
        if flowerbed[i] == 0:
            trai = (i == 0 or flowerbed[i-1] == 0)
            phai = (i == n-1 or flowerbed[i+1] == 0)

            if trai and phai:
                flowerbed[i] = 1
                k -= 1
                if k == 0:
                    return True

    return k <= 0

a = list(map(int, input().split()))
k = int(input())
print(canPlaceFlowers(a, k))
```

```python
#W9A8
def is_path_crossing(moves):
    x = y = 0
    da_di = {(0, 0)}

    for c in moves:
        if c == 'R': x += 1
        elif c == 'L': x -= 1
        elif c == 'U': y += 1
        elif c == 'D': y -= 1

        if (x, y) in da_di:
            return True
        da_di.add((x, y))

    return False

s = input().strip()
print(is_path_crossing(s))
```

```python
#W9A9
def Median(nums):
    a, b, c, d, e = nums
    #print(a,b,c,d,e)
    if a > b: a, b = b, a
    if c > d: c, d = d, c
    if b > d:
        max_nho = d
    else:
        max_nho = b
    if a > c:
        min_lon = a
    else:
        min_lon = c
    if e > min_lon:
        return min_lon
    if e < max_nho:
        return max_nho
    return e
nums = list(map(int,input().split()))
Median(nums)
```

```python
#W9A10
def findRadius(nha,losuoi):
    nha.sort()
    losuoi.sort()

    j = 0
    ban_kinh = 0

    for n in nha:
        while j + 1 < len(losuoi) and abs(losuoi[j+1] - n) <= abs(losuoi[j] - n):
            j += 1
        ban_kinh = max(ban_kinh, abs(losuoi[j] - n))

    return ban_kinh
a=list(map(int,input().split()))
k=list(map(int,input().split()))
print(findRadius(k,a))
```
